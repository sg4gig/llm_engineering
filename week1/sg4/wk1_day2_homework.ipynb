```python
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI

ed = fetch_website_contents("https://edwarddonner.com")
print(ed)
```

```python
system_prompt = """
You are a helpful assistant that will provide a detailed summary of the contents of a website.
You will be given the contents of a website, and then you will be asked to provide a detailed summary.
You should answer the questions based on the contents of the website, and not make up any information.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
user_prompt = """
Here are the contents of a website.
Provide a detailed summary of this website.
If it includes news or announcements, then summarize these too.

"""
user_prompt += ed

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt},
]

print(messages)

OLLAMA_BASE_URL = "http://localhost:11434/v1"

openai = OpenAI()
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

response = ollama.chat.completions.create(model="deepseek-r1:1.5b", messages=messages)

display(Markdown(response.choices[0].message.content))

```
